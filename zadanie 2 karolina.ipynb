```python
import matplotlib.pyplot as plt

miesiace = ["Styczeń", "Luty", "Marzec", "Kwiecień", "Maj", "Czerwiec"]
sprzedaz = [120, 150, 135, 180, 165, 210]

plt.plot(miesiace, sprzedaz, marker="o", color="green")
plt.title("Sprzedaż w kolejnych miesiącach")
plt.xlabel("Miesiąc")
plt.ylabel("Sprzedaż")
plt.xticks(rotation=45)
plt.grid()
plt.tight_layout()
plt.show()
```
